```python
# ============================================================
# PARCIAL TE2 - MINERÍA DE DATOS
# Técnicas de Producción Industrial de Software II
# Alumno: David Salomón Cruz Ramírez
# Carné: 25-0616-2022
# ============================================================

# ============================================================
# 1. INSTALACIÓN E IMPORTACIÓN DE LIBRERÍAS
# ============================================================

!pip install mlxtend -q

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

from mlxtend.frequent_patterns import apriori, association_rules


# ============================================================
# 2. FUNCIÓN PARA MOSTRAR TÍTULOS ORDENADOS
# ============================================================

def mostrar_titulo(texto):
    print("\n" + "=" * 70)
    print(texto)
    print("=" * 70)


# ============================================================
# 3. RUTAS RAW DESDE GITHUB
# ============================================================

ruta_asociacion = "https://raw.githubusercontent.com/DavidCruz1013/Parcial4_DavidCruz_2506162022/main/datos/clave_I_asociacion.csv"
ruta_correlacion = "https://raw.githubusercontent.com/DavidCruz1013/Parcial4_DavidCruz_2506162022/main/datos/clave_I_correlacion.csv"
ruta_agrupacion = "https://raw.githubusercontent.com/DavidCruz1013/Parcial4_DavidCruz_2506162022/main/datos/clave_I_agrupacion.csv"


# ============================================================
# PRIMERA PARTE: RESPUESTAS TEÓRICAS
# ============================================================

mostrar_titulo("PRIMERA PARTE: PRUEBA DE CONOCIMIENTOS")

print("""
1. ¿Qué es minería de datos?
La minería de datos es el proceso de analizar grandes volúmenes de información para descubrir patrones, relaciones, tendencias y comportamientos útiles para la toma de decisiones.

2. ¿Qué es correlación?
La correlación es una medida estadística que permite identificar si dos variables tienen relación entre sí. Esta relación puede ser positiva, negativa o nula.

3. Diferencia entre correlación positiva, negativa y nula:
- Correlación positiva: cuando una variable aumenta, la otra también tiende a aumentar.
- Correlación negativa: cuando una variable aumenta, la otra tiende a disminuir.
- Correlación nula: cuando no existe una relación clara entre las variables.

4. ¿Por qué correlación no significa causalidad?
Porque dos variables pueden estar relacionadas estadísticamente, pero eso no significa que una sea la causa directa de la otra. Puede existir una tercera variable o un factor externo que explique esa relación.

5. ¿Qué son reglas de asociación?
Las reglas de asociación son técnicas de minería de datos que permiten descubrir productos, servicios o elementos que suelen aparecer juntos dentro de un conjunto de transacciones.
""")


# ============================================================
# EJERCICIO 1: ANÁLISIS DE ASOCIACIÓN
# ============================================================

mostrar_titulo("EJERCICIO 1: ANÁLISIS DE ASOCIACIÓN")

df_asociacion = pd.read_csv(ruta_asociacion)

mostrar_titulo("1. DATASET DE ASOCIACIÓN CARGADO CORRECTAMENTE")
display(df_asociacion.head())

mostrar_titulo("2. DIMENSIONES DEL DATASET")
print(f"Filas: {df_asociacion.shape[0]}")
print(f"Columnas: {df_asociacion.shape[1]}")

mostrar_titulo("3. COLUMNAS")
print(df_asociacion.columns.tolist())

mostrar_titulo("4. INFORMACIÓN GENERAL")
df_asociacion.info()

mostrar_titulo("5. VALORES NULOS")
print(df_asociacion.isnull().sum())

mostrar_titulo("6. DUPLICADOS")
print("Duplicados encontrados:", df_asociacion.duplicated().sum())

mostrar_titulo("7. ESTADÍSTICA DESCRIPTIVA")
display(df_asociacion.describe(include="all"))

print("""
Interpretación:
El dataset de asociación contiene registros de ventas por transacción.
Cada fila representa un producto comprado dentro de una transacción.
La columna transaccion_id permite agrupar los productos que fueron comprados juntos.
""")


# Limpieza básica para evitar errores
df_asociacion = df_asociacion.dropna(subset=["transaccion_id", "item", "cantidad"])

# Preparar formato canasta
mostrar_titulo("8. PREPARACIÓN DE TRANSACCIONES")

transacciones = (
    df_asociacion
    .groupby("transaccion_id")["item"]
    .apply(list)
    .tolist()
)

print("Cantidad de transacciones encontradas:", len(transacciones))
print("Ejemplo de una transacción:")
print(transacciones[0])

canasta = (
    df_asociacion
    .groupby(["transaccion_id", "item"])["cantidad"]
    .sum()
    .unstack()
    .fillna(0)
)

canasta = canasta.map(lambda x: 1 if x > 0 else 0)

mostrar_titulo("9. DATOS EN FORMATO CANASTA")
display(canasta.head())

print("""
Interpretación:
Cada fila representa una transacción y cada columna representa un producto.
El valor 1 indica que el producto fue comprado y el valor 0 indica que no fue comprado.
""")


# Aplicar Apriori
mostrar_titulo("10. APLICACIÓN DEL ALGORITMO APRIORI")

productos_frecuentes = apriori(
    canasta,
    min_support=0.03,
    use_colnames=True
)

productos_frecuentes = productos_frecuentes.sort_values(
    by="support",
    ascending=False
)

display(productos_frecuentes.head(10))

print("""
Interpretación:
El soporte indica qué tan frecuente aparece un producto o combinación de productos dentro de todas las transacciones.
""")


# Generar reglas
mostrar_titulo("11. GENERACIÓN DE REGLAS DE ASOCIACIÓN")

reglas = association_rules(
    productos_frecuentes,
    metric="confidence",
    min_threshold=0.30
)

reglas_ordenadas = reglas.sort_values(
    by=["lift", "confidence", "support"],
    ascending=False
)

display(
    reglas_ordenadas[
        ["antecedents", "consequents", "support", "confidence", "lift"]
    ].head(10)
)


# Top 10 reglas
mostrar_titulo("12. LAS 10 REGLAS MÁS RELEVANTES")

top10_reglas = reglas_ordenadas.head(10).copy()

for indice, fila in top10_reglas.iterrows():
    antecedente = list(fila["antecedents"])
    consecuente = list(fila["consequents"])

    print("\nRegla:")
    print(f"Si compra {antecedente}, también puede comprar {consecuente}")
    print(f"Soporte: {fila['support']:.3f}")
    print(f"Confianza: {fila['confidence']:.3f}")
    print(f"Lift: {fila['lift']:.3f}")


# Interpretación de 5 reglas
mostrar_titulo("13. INTERPRETACIÓN DE 5 REGLAS")

for numero, (_, fila) in enumerate(top10_reglas.head(5).iterrows(), start=1):
    antecedente = ", ".join(list(fila["antecedents"]))
    consecuente = ", ".join(list(fila["consequents"]))

    print(f"""
Regla {numero}:
Cuando un cliente compra {antecedente}, existe una probabilidad importante de que también compre {consecuente}.
Esto puede ser útil para crear promociones, combos o sugerencias de compra.
Confianza: {fila['confidence']:.2f}
Lift: {fila['lift']:.2f}
""")


# Recomendaciones
mostrar_titulo("14. RECOMENDACIONES COMERCIALES")

print("""
1. Crear combos con los productos que aparecen relacionados en las reglas más fuertes.

2. Colocar productos relacionados cerca dentro de la tienda para facilitar la compra cruzada.

3. Usar las reglas obtenidas para mostrar recomendaciones automáticas en una aplicación o plataforma de ventas.

4. Diseñar promociones específicas para productos que tienen alto lift, porque presentan una relación fuerte de compra conjunta.
""")


# ============================================================
# EJERCICIO 2: ANÁLISIS DE CORRELACIÓN
# ============================================================

mostrar_titulo("EJERCICIO 2: ANÁLISIS DE CORRELACIÓN")

df_correlacion = pd.read_csv(ruta_correlacion)

mostrar_titulo("15. DATASET DE CORRELACIÓN CARGADO")
display(df_correlacion.head())

mostrar_titulo("16. DIMENSIONES")
print(f"Filas: {df_correlacion.shape[0]}")
print(f"Columnas: {df_correlacion.shape[1]}")

mostrar_titulo("17. COLUMNAS")
print(df_correlacion.columns.tolist())

mostrar_titulo("18. INFORMACIÓN GENERAL")
df_correlacion.info()

mostrar_titulo("19. VALORES NULOS")
print(df_correlacion.isnull().sum())

mostrar_titulo("20. DUPLICADOS")
print("Duplicados encontrados:", df_correlacion.duplicated().sum())

mostrar_titulo("21. ESTADÍSTICA DESCRIPTIVA")
display(df_correlacion.describe())


# Variables numéricas
variables_numericas_corr = df_correlacion.select_dtypes(include=["int64", "float64"]).copy()

mostrar_titulo("22. VARIABLES NUMÉRICAS IDENTIFICADAS")
print(variables_numericas_corr.columns.tolist())


# Limpieza de valores nulos
mostrar_titulo("23. LIMPIEZA DE VALORES NULOS EN CORRELACIÓN")
print("Valores nulos antes:")
print(variables_numericas_corr.isnull().sum())

variables_numericas_corr = variables_numericas_corr.fillna(variables_numericas_corr.mean())

print("\nValores nulos después:")
print(variables_numericas_corr.isnull().sum())


# Revisión de atípicos
mostrar_titulo("24. REVISIÓN DE VALORES ATÍPICOS")

for variable in variables_numericas_corr.columns:
    promedio = variables_numericas_corr[variable].mean()
    desviacion = variables_numericas_corr[variable].std()

    limite_bajo = promedio - (2 * desviacion)
    limite_alto = promedio + (2 * desviacion)

    datos_atipicos = variables_numericas_corr[
        (variables_numericas_corr[variable] < limite_bajo) |
        (variables_numericas_corr[variable] > limite_alto)
    ]

    print(f"\nVariable analizada: {variable}")
    print(f"Promedio: {promedio:.2f}")
    print(f"Desviación estándar: {desviacion:.2f}")
    print(f"Límite inferior: {limite_bajo:.2f}")
    print(f"Límite superior: {limite_alto:.2f}")
    print(f"Cantidad de valores atípicos: {len(datos_atipicos)}")


# Matriz de correlación
mostrar_titulo("25. MATRIZ DE CORRELACIÓN")

matriz_correlacion = variables_numericas_corr.corr()

display(matriz_correlacion.round(3))


# Mapa de calor
mostrar_titulo("26. MAPA DE CALOR")

plt.figure(figsize=(12, 8))
plt.imshow(matriz_correlacion, cmap="coolwarm")
plt.colorbar(label="Nivel de correlación")
plt.xticks(range(len(matriz_correlacion.columns)), matriz_correlacion.columns, rotation=90)
plt.yticks(range(len(matriz_correlacion.columns)), matriz_correlacion.columns)
plt.title("Mapa de calor de correlación")

for i in range(len(matriz_correlacion.columns)):
    for j in range(len(matriz_correlacion.columns)):
        plt.text(j, i, round(matriz_correlacion.iloc[i, j], 2),
                 ha="center", va="center")

plt.tight_layout()
plt.show()


# Correlaciones positivas y negativas
mostrar_titulo("27. CORRELACIONES POSITIVAS Y NEGATIVAS")

pares_correlacion = (
    matriz_correlacion
    .where(np.triu(np.ones(matriz_correlacion.shape), k=1).astype(bool))
    .stack()
    .reset_index()
)

pares_correlacion.columns = ["Variable_1", "Variable_2", "Correlacion"]

correlaciones_positivas = pares_correlacion.sort_values(
    by="Correlacion",
    ascending=False
)

correlaciones_negativas = pares_correlacion.sort_values(
    by="Correlacion",
    ascending=True
)

print("\nCorrelaciones positivas más importantes:")
display(correlaciones_positivas.head(5))

print("\nCorrelaciones negativas más importantes:")
display(correlaciones_negativas.head(5))


# Explicar relaciones
mostrar_titulo("28. EXPLICACIÓN DE 4 RELACIONES")

for numero, (_, fila) in enumerate(correlaciones_positivas.head(2).iterrows(), start=1):
    print(f"""
Relación positiva {numero}:
La variable {fila['Variable_1']} tiene una correlación positiva de {fila['Correlacion']:.2f} con {fila['Variable_2']}.
Esto indica que cuando una variable aumenta, la otra también tiende a aumentar.
""")

for numero, (_, fila) in enumerate(correlaciones_negativas.head(2).iterrows(), start=1):
    print(f"""
Relación negativa {numero}:
La variable {fila['Variable_1']} tiene una correlación negativa de {fila['Correlacion']:.2f} con {fila['Variable_2']}.
Esto indica que cuando una variable aumenta, la otra tiende a disminuir.
""")


mostrar_titulo("29. CONCLUSIÓN DE CORRELACIÓN")

print("""
El análisis de correlación permitió identificar relaciones importantes entre las variables numéricas del dataset.
Estas relaciones pueden ayudar a la empresa a comprender mejor el comportamiento de sus clientes.

Por ejemplo, si se observa una relación negativa entre reclamos y satisfacción, se puede interpretar que a mayor cantidad de reclamos, menor satisfacción del cliente.
También, si existe una relación positiva entre frecuencia de compra y consumo total, la empresa puede enfocar estrategias de fidelización en los clientes más frecuentes.
""")


# ============================================================
# EJERCICIO 3: AGRUPACIÓN O CLUSTERING
# ============================================================

mostrar_titulo("EJERCICIO 3: AGRUPACIÓN O CLUSTERING")

df_agrupacion = pd.read_csv(ruta_agrupacion)

mostrar_titulo("30. DATASET DE AGRUPACIÓN CARGADO")
display(df_agrupacion.head())

mostrar_titulo("31. DIMENSIONES")
print(f"Filas: {df_agrupacion.shape[0]}")
print(f"Columnas: {df_agrupacion.shape[1]}")

mostrar_titulo("32. COLUMNAS")
print(df_agrupacion.columns.tolist())

mostrar_titulo("33. INFORMACIÓN GENERAL")
df_agrupacion.info()

mostrar_titulo("34. VALORES NULOS")
print(df_agrupacion.isnull().sum())

mostrar_titulo("35. DUPLICADOS")
print("Duplicados encontrados:", df_agrupacion.duplicated().sum())

mostrar_titulo("36. ESTADÍSTICA DESCRIPTIVA")
display(df_agrupacion.describe())


# Variables para clustering
variables_cluster = [
    "edad",
    "ingresos",
    "frecuencia_uso",
    "gasto_promedio",
    "satisfaccion",
    "reclamos",
    "antiguedad_meses"
]

datos_cluster = df_agrupacion[variables_cluster].copy()

mostrar_titulo("37. VARIABLES SELECCIONADAS PARA CLUSTERING")
print(variables_cluster)

print("""
Justificación:
Se seleccionaron variables relacionadas con el comportamiento y valor del cliente:
edad, ingresos, frecuencia de uso, gasto promedio, satisfacción, reclamos y antigüedad.
Estas variables permiten segmentar clientes según consumo, lealtad, experiencia y nivel de relación con la empresa.
""")


# Limpieza de NaN en clustering
mostrar_titulo("38. LIMPIEZA DE VALORES NULOS EN CLUSTERING")

print("Valores nulos antes:")
print(datos_cluster.isnull().sum())

datos_cluster = datos_cluster.fillna(datos_cluster.mean())

print("\nValores nulos después:")
print(datos_cluster.isnull().sum())


# Normalización
mostrar_titulo("39. NORMALIZACIÓN DE DATOS")

escalador = StandardScaler()
datos_escalados = escalador.fit_transform(datos_cluster)

print("Datos escalados correctamente.")


# Método del codo
mostrar_titulo("40. MÉTODO DEL CODO")

inercias = []

for k in range(1, 11):
    modelo = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    modelo.fit(datos_escalados)
    inercias.append(modelo.inertia_)

plt.figure(figsize=(8, 6))
plt.plot(range(1, 11), inercias, marker="o")
plt.xlabel("Número de clústeres")
plt.ylabel("Inercia")
plt.title("Método del Codo para seleccionar K")
plt.grid(True)
plt.show()

print("""
Interpretación:
El método del codo permite observar en qué punto la reducción de la inercia deja de ser tan fuerte.
Para este análisis se utilizarán 3 clústeres, porque permite una segmentación clara y fácil de interpretar.
""")


# Aplicar K-Means
mostrar_titulo("41. APLICACIÓN DEL MODELO K-MEANS")

kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df_agrupacion["Cluster"] = kmeans.fit_predict(datos_escalados)

mostrar_titulo("42. DATASET CON COLUMNA CLUSTER")
display(df_agrupacion.head())


# Centroides
mostrar_titulo("43. CENTROIDES DEL MODELO")

centroides = (
    df_agrupacion
    .groupby("Cluster")[variables_cluster]
    .mean()
    .round(2)
)

display(centroides)


# Cantidad por cluster
mostrar_titulo("44. CANTIDAD DE CLIENTES POR CLUSTER")
print(df_agrupacion["Cluster"].value_counts().sort_index())


# Visualizaciones
mostrar_titulo("45. VISUALIZACIÓN DE CLUSTERS")

plt.figure(figsize=(8, 6))
plt.scatter(
    df_agrupacion["ingresos"],
    df_agrupacion["gasto_promedio"],
    c=df_agrupacion["Cluster"]
)
plt.xlabel("Ingresos")
plt.ylabel("Gasto promedio")
plt.title("Clusters de clientes: Ingresos vs Gasto promedio")
plt.grid(True)
plt.show()


plt.figure(figsize=(8, 6))
plt.scatter(
    df_agrupacion["satisfaccion"],
    df_agrupacion["reclamos"],
    c=df_agrupacion["Cluster"]
)
plt.xlabel("Satisfacción")
plt.ylabel("Reclamos")
plt.title("Clusters de clientes: Satisfacción vs Reclamos")
plt.grid(True)
plt.show()


plt.figure(figsize=(8, 6))
plt.scatter(
    df_agrupacion["frecuencia_uso"],
    df_agrupacion["gasto_promedio"],
    c=df_agrupacion["Cluster"]
)
plt.xlabel("Frecuencia de uso")
plt.ylabel("Gasto promedio")
plt.title("Clusters de clientes: Frecuencia de uso vs Gasto promedio")
plt.grid(True)
plt.show()


# Interpretación por cluster
mostrar_titulo("46. INTERPRETACIÓN DE CADA CLUSTER")

for cluster in sorted(df_agrupacion["Cluster"].unique()):
    datos = centroides.loc[cluster]

    print(f"""
Cluster {cluster}:
Edad promedio: {datos['edad']:.2f}
Ingresos promedio: {datos['ingresos']:.2f}
Frecuencia de uso promedio: {datos['frecuencia_uso']:.2f}
Gasto promedio: {datos['gasto_promedio']:.2f}
Satisfacción promedio: {datos['satisfaccion']:.2f}
Reclamos promedio: {datos['reclamos']:.2f}
Antigüedad promedio en meses: {datos['antiguedad_meses']:.2f}
""")


print("""
Interpretación general:
Los clústeres permiten identificar grupos de clientes con características similares.
Algunos grupos pueden representar clientes de alto valor, otros clientes con bajo uso del servicio y otros clientes que requieren atención por bajo nivel de satisfacción o muchos reclamos.
""")


# Acciones de negocio
mostrar_titulo("47. ACCIONES DE NEGOCIO POR CLUSTER")

print("""
Cluster 0:
Se recomienda aplicar estrategias de fidelización, promociones exclusivas y beneficios por permanencia.

Cluster 1:
Se recomienda motivar el uso del servicio mediante descuentos, campañas personalizadas y recordatorios por medio de la aplicación.

Cluster 2:
Se recomienda mejorar la atención al cliente, revisar reclamos frecuentes y ofrecer seguimiento personalizado para aumentar la satisfacción.
""")


# Exportar dataset final
mostrar_titulo("48. EXPORTACIÓN DEL DATASET FINAL")

nombre_archivo_cluster = "clave_I_agrupacion_con_cluster.csv"

df_agrupacion.to_csv(nombre_archivo_cluster, index=False)

print("Archivo exportado correctamente:")
print(nombre_archivo_cluster)


# Descargar archivo final en Colab
try:
    from google.colab import files
    files.download(nombre_archivo_cluster)
except:
    print("Si no estás en Google Colab, el archivo quedó guardado en tu entorno local.")


# ============================================================
# CONCLUSIONES GENERALES
# ============================================================

mostrar_titulo("49. CONCLUSIONES GENERALES")

print("""
1. En el análisis de asociación se identificaron productos que suelen comprarse juntos, lo cual puede ayudar a crear combos, promociones y estrategias de venta cruzada.

2. En el análisis de correlación se identificaron relaciones positivas y negativas entre variables de clientes, lo cual permite comprender mejor su comportamiento.

3. En el análisis de clustering se agruparon clientes con características similares, permitiendo segmentar la cartera de clientes de manera más estratégica.

4. La columna Cluster agregada al dataset final permite que la empresa clasifique a sus clientes y tome decisiones diferenciadas para cada grupo.

5. Las técnicas utilizadas en este parcial demuestran cómo la minería de datos puede apoyar la toma de decisiones empresariales mediante análisis, gráficos e interpretación de resultados.
""")

mostrar_titulo("50. PARCIAL FINALIZADO CORRECTAMENTE")
print("Proceso completado.")
```
